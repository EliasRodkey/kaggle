## JOINs

There are several types of JOINs:
- **INNER JOIN**: Returns records that have matching values in both tables

```sql
SELECT o.Name AS Owner_Name, p.Name AS Pet_Name
FROM `bigquery-public-data.pet_records.owners` AS o
INNER JOIN `bigquery-public-data.pet_records.pets` AS p
    ON p.ID = o.pet_id
```

- **LEFT JOIN**: Returns all records from the left table, and the matched records from the right table

```sql
SELECT o.Name AS Owner_Name, p.Name AS Pet_Name
FROM `bigquery-public-data.pet_records.owners` AS o
LEFT JOIN `bigquery-public-data.pet_records.pets` AS p
    ON p.ID = o.pet_id
```
- **RIGHT JOIN**: Returns all records from the right table, and the matched records from the left table

```sql
SELECT o.Name AS Owner_Name, p.Name AS Pet_Name
FROM `bigquery-public-data.pet_records.owners` AS o
RIGHT JOIN `bigquery-public-data.pet_records.pets` AS p
    ON p.ID = o.pet_id
```

- For left and right joins, the table before the JOIN keyword is the left table, and the table after the JOIN keyword is the right table

- **FULL JOIN**: Returns all records from both tables

```sql
SELECT o.Name AS Owner_Name, p.Name AS Pet_Name
FROM `bigquery-public-data.pet_records.owners` AS o
FULL JOIN `bigquery-public-data.pet_records.pets` AS p
    ON p.ID = o.pet_id
```

## UNIONS

UNIONS combine the results of two or more SELECT statements into a single result set.
This results in a concatenation of the two tables.

```sql
SELECT Age FROM `bigquery-public-data.pet_records.pets`
UNION All
SELECT Age FROM `bigquery-public-data.pet_records.owners`;
```

- Data types must match between the two tables
- UNION removes duplicates, UNION ALL keeps them, UNION DISTINCT removes duplicates
