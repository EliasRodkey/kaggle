# Select, From, & Where

SELECT, FROM, and WHERE are 3 fundamental SQL keywords that allow the retrieval of data from specific columns that matches the given constraints. Can also be used to fetch the entire table with SELECT * FROM table_name.

General Format:

```sql
SELECT column_name(s)
FROM table_name
WHERE parameters
```

* SELECT: can be comma seperated column names, * for all columns
* FROM: the name of the table the data is being pulled from. (for example: FROM `bigquery-public-data.openaq.global_air_quality`. The argument passed to FROM to bigquery client can be surrounded in backticks (``) but not ' or ")
* WHERE: can be comma seperated conditional statements
- column_name (i.e. Animal) 
- equality_operator (=, <, >, between etc.) 
- value ('Cat', 9, etc.)

### Working with Big Datasets

Some datasets on BigQuery are HUGE. To avoid running through the free limit or overloading your system memory you can preview the size of the query return by creating a QueryJobConfig and setting dry_run to True.

```python
# Create a QueryJobConfig object to estimate size of query without running it
dry_run_config = bigquery.QueryJobConfig(dry_run=True)

# API request - dry run query to estimate costs
dry_run_query_job = client.query(query, job_config=dry_run_config)

print("This query will process {} bytes.".format(dry_run_query_job.total_bytes_processed))
```


Or alternatively you can create a QueryJobConfig that will produce an error if a certain query size is exceeded:

```python
# Only run the query if it's less than 1 MB
ONE_MB = 1000*1000
safe_config = bigquery.QueryJobConfig(maximum_bytes_billed=ONE_MB)

# Set up the query (will only run if it's less than 1 MB)
safe_query_job = client.query(query, job_config=safe_config)

# API request - try to run the query, and return a pandas DataFrame
safe_query_job.to_dataframe()
```
