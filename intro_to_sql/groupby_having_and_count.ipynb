# Group By, Having, and Count

### COUNT

COUNT is an aggregate function (like avg(), sum(), min()). COUNT is used in conjunction with SELECT to return the number of items in a column.

```sql
SELECT COUNT(ID) FROM table_name
```

Above returns the number of items in the ID column from the table table_name.

### GROUP BY

GROUP BY can be used in conjunction with an aggregate function to give the count, min, max, average of values associated with categories.

```sql
SELECT category, AVG(amount) FROM transactions_table GROUP BY category
```

Above returns the average amount of transactions by category.

### HAVING

HAVING is used in combination with GROUP BY to ignore groups that don't meet a given criteria.

```sql
SELECT Animal, COUNT(ID)
FROM table_name
GROUP BY Animal
HAVING COUNT(ID) > 1
```

Above returns counts of all animals who have counts greater than 1.

### Aliasing

Aggregate functions like COUNT will return a strange column name (something like f0__). This can be renamed inside the SQL query itself using the AS keyword.

```sql
SELECT Animal, COUNT(ID) AS NumAnimals
FROM table_name
GROUP BY Animal
HAVING COUNT(ID) > 1 
```
