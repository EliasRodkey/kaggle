# Getting Started with SQL and BigQuery

SQL or Structured Query Language is the programming language of databases.
BigQuery is a web service that allows you to access large datasets from the internet.

````python
from google.cloud import bigquery
````

After importing bigquery, we need to create a Client object which will be the connection we use to retrieve data.

````python
client = bigquery.Client()

# This doesn't work on my computer since you need to have configured you google cloud credentials and I don't want to do that on my work PC:
# https://docs.cloud.google.com/docs/authentication/set-up-adc-local-dev-environment#local-user-cred
````

Next we use the dataset and get_dataset functions to retrieve the data we are looking for.

````python
# Construct a reference to the "hacker_news" dataset
dataset_ref = client.dataset("hacker_news", project="bigquery-public-data")

# API request - fetch the dataset
dataset = client.get_dataset(dataset_ref)
````

Each dataset is a collection of tables. Use list_tables() to list the tables available in the dataset.

````python
# List all the tables in the "hacker_news" dataset
tables = list(client.list_tables(dataset))

# Print names of all tables in the dataset (there are four!)
for table in tables:  
    print(table.table_id)
````

Tables can be fetched from within the dataset in a similar way to how the dataset was retrieved originally.

````python
# Construct a reference to the "full" table
table_ref = dataset_ref.table("full")

# API request - fetch the table
table = client.get_table(table_ref)
````

The client is our connection to the bigquery service and holds connections to the different projects. A porject is a collection of datasets, and a dataset contains one or more tables that have the actual data.

### Schema

````python
# Print information on all the columns in the "full" table in the "hacker_news" dataset
table.schema
````

Each SchemaField object tells us about a specific column (field). 

```python
SchemaField('by', 'string', 'NULLABLE', "The username of the item's author.",())
```

This tells us that:
* the field (or column) is called "by",
* the data in this field is strings,
* NULL values are allowed, and
* it contains the usernames corresponding to each item's author.

We can use the list_rows() method to check just the first five lines of of the full table to make sure this is right. (Sometimes databases have outdated descriptions, so it's good to check.) This returns a BigQuery RowIterator object that can quickly be converted to a pandas DataFrame with the to_dataframe() method.

````python
# Preview the first five lines of the "full" table
client.list_rows(table, max_results=5).to_dataframe()
````

````python
# Use selected_fields to specify columns to view using list_rows
# Preview the first five entries in the "by" column of the "full" table
client.list_rows(table, selected_fields=table.schema[:1], max_results=5).to_dataframe()
````
