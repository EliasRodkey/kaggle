# Scatter Plots

### First Load and Examine the Data

```python
import pandas as pd
pd.plotting.register_matplotlib_converters()
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
print("Setup Complete")
```

```python
# Path of the file to read
insurance_filepath = "data/insurance.csv"

# Read the file into a variable insurance_data
insurance_data = pd.read_csv(insurance_filepath)
```

```python
insurance_data.head()
```

### Use sns.scatterplot()

```python
# Create a simple scatter plot
sns.scatterplot(x=insurance_data['bmi'], y=insurance_data['charges'])
```

```python
# Use .regplot() to include a regression line that best fits the data
sns.regplot(x=insurance_data['bmi'], y=insurance_data['charges'])
```

### Hues and Multiple Regression Lines

```python
# Add a hue to the scatter plot to track 3 variables visually
sns.scatterplot(x=insurance_data['bmi'], y=insurance_data['charges'], hue=insurance_data['smoker'])
```

```python
# Add a second regression line using .lmplot()
sns.lmplot(x="bmi", y="charges", hue="smoker", data=insurance_data)
```

### Categorical Scatter Plots

Used to plot categorical variables (i.e. true, false)

```python
sns.swarmplot(x=insurance_data['smoker'],
              y=insurance_data['charges'])
```
