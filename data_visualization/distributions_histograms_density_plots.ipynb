# Distributions: Histograms and Density Plots

### Setup Code

```python
import pandas as pd
pd.plotting.register_matplotlib_converters()
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
print("Setup Complete")
```

### Load and Examine Data

```python
# Path of the file to read
iris_filepath = "data/iris.csv"

# Read the file into a variable iris_data
iris_data = pd.read_csv(iris_filepath, index_col="Id")

# Print the first 5 rows of the data
iris_data.head()
```

### Histogram Plots

```python
# Normal histogram 
sns.histplot(iris_data['Petal Length (cm)'])
```

```python
# Color coded histograms for each species
sns.histplot(data=iris_data, x='Petal Length (cm)', hue='Species')

# Add title
plt.title("Histogram of Petal Lengths, by Species")
```

```python
# KDE (kernel density estimate) plot. Smoothed histogram
sns.kdeplot(data=iris_data['Petal Length (cm)'], fill=True)
```

```python
# Color coded KDE plots for each species
sns.kdeplot(data=iris_data, x='Petal Length (cm)', hue='Species', shade=True)

# Add title
plt.title("Distribution of Petal Lengths, by Species")
```

### 2D Kernel Density Estimate Plot
uses jointplot()

```python
# 2D KDE plot, show the density of two features and creates topographical image.
sns.jointplot(x=iris_data['Petal Length (cm)'], y=iris_data['Sepal Width (cm)'], kind="kde")
```
