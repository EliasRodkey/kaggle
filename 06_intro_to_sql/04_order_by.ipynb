# Order By

ORDER BY is generally the last clause in the query since it only determines how the data is returned. It sorts the results returned by the query.

```sql
SELECT ID, Name, Animal
FROM table_name
ORDER BY ID
```

Also works for columns with text and sorts them alphabetically.

Order can be reversed using the DESC argument.


```sql
SELECT ID, Name, Animal
FROM table_name
ORDER BY ID DESC
```

## EXTRACT

Allows you to look at a part of a date or datetime.

```sql
SELECT Name, EXTRACT(DAY from Date) AS Day
FROM `bigquery-public-data.pet_records.pets_with_date`
```

SQL is very smart with date manipulation, it can also return things like which week of the year the date falls in and what day of the week it is.

```sql
SELECT Name, EXTRACT(WEEK FROM Date) AS Week
FROM `bigquery-public-data.pet_records.pets_with_date`
```
