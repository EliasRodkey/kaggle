# AS & WITH
organizing queries

### AS
- AS can be used to rename columns or tables
- return the same information but with cleaner and clearer column names
- used on its iwn it is a simple way to clean up and organize columns

### WITH
- WITH and AS together when combined are refered to as a "common table expession" or CTE
- CTEs are temporary tables that can be used in a query
- CTEs are useful for organizing queries and making them more readable
- CTEs are also useful for recursive queries

### CTE Example
- select specific subset of data from the table at the top and name it 'Seniors'
- 'Seniors' can now be referenced in the outer query and data can be selected from it

```sql
WITH Seniors AS
(
    SELECT ID, Name
    FROM `bigquery-public-data.pet_records.pets`
    WHERE Years_old > 5
)
SELECT ID
FROM Seniors
```

- it is possible to do this without a CTE but it is less readable and more verbose
- always create the CTE first and then reference it in the outer query
